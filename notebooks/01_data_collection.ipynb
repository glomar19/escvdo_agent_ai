```python
#Download the images from ESCVDO
import pandas as pd
from parse_saved_page import parse_saved_page

colecciones = ["FW25", "PF25", "PF26", "PS26", "SS26"]
tablas = []

for c in colecciones:
    rows = parse_saved_page(
        str(ROOT / f"ESCVDO_{c}.html"),
        str(ROOT / f"images/escvdo_{c.lower()}"),
        site_url="https://www.escvdo.com",
    )
    t = pd.DataFrame(rows)
    t["brand"] = "escvdo"
    t["collection"] = c
    t["folder"] = f"escvdo_{c.lower()}"
    tablas.append(t)

escvdo = pd.concat(tablas, ignore_index=True)
escvdo.to_csv(ROOT / "data/escvdo_all.csv", index=False)

print(len(escvdo))
escvdo.groupby("collection")["category"].value_counts()
```
